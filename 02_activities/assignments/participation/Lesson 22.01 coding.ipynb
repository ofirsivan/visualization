```python
np.random.seed(613) #setting a random seed for reproducability
x = np.arange(50)  # random 50 integers [0-49]
y1 = np.random.randint(0, 100,50)  # determine y1 and y2
y2 = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot (x,y2)
plt.show ()
```

```python
#annotating with arrows

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, marker= "^", color="purple", label= "Person A")
ax.scatter (x,y2, marker= "*", color="gold", label= "Person B")
ax.legend (loc = "lower left", shadow= True, frameon= True, ncol=1, fontsize=8, bbox_to_anchor = (0,1), prop={"family":"serif","size":8})
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94), arrowprops=dict(facecolor='green'))

plt.show ()
```

```python
#annotating with arrows

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, marker= "^", color="purple", label= "Person A")
ax.scatter (x,y2, marker= "*", color="gold", label= "Person B")
ax.legend (loc = "lower left", shadow= True, frameon= True, ncol=1, fontsize=8, bbox_to_anchor = (0,1), prop={"family":"serif","size":8})
ax.annotate ("This is important!", xy=(10, 95), xytext=(20, 94), arrowprops = dict(arrowstyle = "wedge",color = "deeppink"))
plt.show ()
```

```python
#modify lables on axes 

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, marker= "^", color="purple", label= "Person A")
ax.scatter (x,y2, marker= "*", color="gold", label= "Person B")
ax.legend (loc = "lower left", shadow= True, frameon= True, ncol=1, fontsize=8, bbox_to_anchor = (0,1), prop={"family":"serif","size":8})
ax.yaxis.set_major_locator(plt.NullLocator()) 
ax.xaxis.set_major_formatter(plt.NullFormatter())
plt.show ()
```

```python
#decide how many lables we want on our x axis
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, marker= "^", color="purple", label= "Person A")
ax.scatter (x,y2, marker= "*", color="gold", label= "Person B")
ax.legend (loc = "lower left", shadow= True, frameon= True, ncol=1, fontsize=8, bbox_to_anchor = (0,1), prop={"family":"serif","size":8})
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
plt.show ()
```

```python
#modify multiple locators 
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, marker= "^", color="purple", label= "Person A")
ax.scatter (x,y2, marker= "*", color="gold", label= "Person B")
ax.legend (loc = "lower left", shadow= True, frameon= True, ncol=1, fontsize=8, bbox_to_anchor = (0,1), prop={"family":"serif","size":8})
ax.xaxis.set_major_locator(plt.MaxNLocator(5))
ax.yaxis.set_major_locator(plt.MaxNLocator(5))
plt.show ()
```

```python
#with intervals with multiple locators 
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, marker= "^", color="purple", label= "Person A")
ax.scatter (x,y2, marker= "*", color="gold", label= "Person B")
ax.legend (loc = "lower left", shadow= True, frameon= True, ncol=1, fontsize=8, bbox_to_anchor = (0,1), prop={"family":"serif","size":8})
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
# with roatation
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, marker= "^", color="purple", label= "Person A")
ax.scatter (x,y2, marker= "*", color="gold", label= "Person B")
ax.legend (loc = "lower left", shadow= True, frameon= True, ncol=1, fontsize=8, bbox_to_anchor = (0,1), prop={"family":"serif","size":8})
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.xticks(rotation=45, ha='right')
```
