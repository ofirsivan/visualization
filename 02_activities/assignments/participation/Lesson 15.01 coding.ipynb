```python
#First, we will create a random seed for reproducability,and then we randomally generate
# X with 50 (int 0-49) numbers, and y (0-100) with 50 numbers
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
# we specify the objects fig and ax 
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
# now we will create a bar plot for each observation [here x values are still descrete]
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)

```

```python
# After plotting the bar-chart, I will create a line-plot (connecting datapoints with lines) in order to show trends in data [e.g., stockmarket]
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)

```

```python
#activity - create a histogram - we do not need to plot the X axis for histogram only the y [X datasype is continouus]
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
ax.hist (y,edgecolor = 'black')
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {'family':'sans-serif','color':'purple','size':20}
font2 = {'family':'monospace','color':'green','size':14}
font3 = {'family':'monospace','color':'red','size':14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font3)
fig.tight_layout()
```

```python
#we can modify our datapoints in the sactter plot
```

```python
# and now with green diamonds! 
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y, marker='D', color = "green")   
plt.show()
```

```python
# then we plot connection between datapoints
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='D', color = "deeppink", linestyle = '--', linewidth = 2)   
plt.show()
```

```python
# then we plot connection between datapoints.changing marker's face and edge, shape 
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='^', markersize= 7, color = "green", linestyle = '--', linewidth = 2, markeredgecolor= 'purple', markerfacecolor='gold')   
plt.show()
```

```python
# then we plot connection between datapoints.changing marker's face and edge, shape 
fig, ax = plt.subplots(figsize=(5, 3))
ax.grid(axis = 'y', color= 'blue')
ax.plot(x, y, marker='^', markersize= 7, color = "green", linestyle = '--', linewidth = 2, markeredgecolor= 'purple', markerfacecolor='gold')   
plt.show()
```
